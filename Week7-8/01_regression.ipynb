### 1. Mathematical Background

- **Linear Regression Loss (Ordinary Least Squares)**:
$$J(\theta) = \frac{1}{2m} \sum_{i=1}^{m} \left( h_\theta(x^{(i)}) - y^{(i)} \right)^2$$

- **Polynomial Regression**:
$$y = \theta_0 + \theta_1 x + \theta_2 x^2 + \dots + \theta_d x^d + \epsilon$$

- **Ridge Regression (L2 Penalty)**:
$$J_{Ridge}(\theta) = J(\theta) + \alpha \sum_{j=1}^{n} \theta_j^2$$

- **Lasso Regression (L1 Penalty & Feature Selection)**:
$$J_{Lasso}(\theta) = J(\theta) + \alpha \sum_{j=1}^{n} |\theta_j|$$
